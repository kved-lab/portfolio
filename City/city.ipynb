```python
import pandas as pd
import pymysql
from sqlalchemy import create_engine

# Charger le fichier Excel
file_path = "communes-departement-region.csv"  # Remplace par le chemin de ton fichier Excel
#df = pd.read_excel(file_path)
df = pd.read_csv(file_path)


# Connexion à la base de données MariaDB en utilisant SQLAlchemy
connection = pymysql.connect(host='localhost',
                             user='fred',
                             password='Canard01',
                             port=3306,
                             db='city') 

# Nom de la table
table_name = "city"  # Remplace par le nom de ta table

# Créer la table si elle n'existe pas
with connection.cursor() as cursor:
    # Construire une requête SQL pour créer une table avec les colonnes du DataFrame
    columns = ", ".join([f"`{col}` TEXT" for col in df.columns])  # Assumer que toutes les colonnes sont de type TEXTE
    create_table_query = f"CREATE TABLE IF NOT EXISTS `{table_name}` ({columns})"
    cursor.execute(create_table_query)

engine = create_engine('mariadb+pymysql://fred:Canard01@localhost/city?charset=utf8mb4', echo=False)
conn=engine.connect()

# Insérer les données du DataFrame dans la table
df.to_sql(table_name, con=conn, if_exists='append', index=False)

print(f"Les données ont été insérées dans la table '{table_name}' avec succès.")
```
